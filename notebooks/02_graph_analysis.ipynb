# 2. Communication graph

Structure of the graph the models actually see: how many endpoints, how
many ordered pairs, and how heavily parallel the edges are. The parallel
multiplicity is the number that justifies keeping one edge per flow record
rather than collapsing pairs.

```python
import sys, json; sys.path.insert(0, '..')
import numpy as np, pandas as pd
from src.utils import DATA_METADATA, RESULTS_RAW, RESULTS_TABLES, collect_runs, load_json
```

```python
ctx = [load_json(p) for p in sorted(DATA_METADATA.glob('context__*.json'))]
g = pd.DataFrame([{**{k: c[k] for k in ('representation','task','protocol','seed')}, **c['graph_statistics']} for c in ctx])
g[['representation','protocol','n_nodes','n_edges_flow_instances','n_unique_ordered_pairs','parallel_edge_multiplicity_mean','parallel_edge_multiplicity_max','graph_density_ordered']].drop_duplicates()
```
