# 3. Baselines

How far the tabular models get on their own. If these are already at the
ceiling on every metric that matters, the graph has nothing to add and the
paper should say so.

```python
import sys, json; sys.path.insert(0, '..')
import numpy as np, pandas as pd
from src.utils import DATA_METADATA, RESULTS_RAW, RESULTS_TABLES, collect_runs, load_json
```

```python
runs = collect_runs()
base = runs[(runs.experiment=='main') & runs.model.isin(['RandomForest','XGBoost','SVM','KNN','GA-EBT','MLP'])]
base.groupby(['model','representation','task','protocol'])[['accuracy','f1_macro','mcc','pr_auc','fpr','minority_class_recall']].agg(['mean','std']).round(4)
```

## Accuracy is saturated; the other metrics are not

Plot the spread of each metric across models. A metric whose range across
*all* models is smaller than its seed-to-seed standard deviation cannot
discriminate between methods and must not carry the argument.

```python
for m in ['accuracy','f1_macro','mcc','pr_auc','fpr','minority_class_recall']:
    s = base.groupby('model')[m].mean()
    print(f'{m:24s} range across models {s.max()-s.min():.5f}   mean seed s.d. {base.groupby("model")[m].std().mean():.5f}')
```
